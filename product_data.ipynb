# Product Search

```python

def get_product_search_results(query, country_code='IN', api_key=OC_API_KEY):
    response = requests.get(
        "https://amazon-scraper-api.omkar.cloud/amazon/search",
        params={"query": query, "country_code": country_code},
        headers={"API-Key": api_key}
    )
    return response.json()

# response = requests.get(
#     "https://amazon-scraper-api.omkar.cloud/amazon/search",
#     params={"query": "Hoodie", "country_code": "IN"},
#     headers={"API-Key": OC_API_KEY}
# )

# print(response.json())
```

# Category Search

```python
def get_category_products(category_id, country_code="IN", api_key=OC_API_KEY):
    response = requests.get(
        "https://amazon-scraper-api.omkar.cloud/amazon/products/category",
        params={"category_id": category_id, "country_code": country_code},
        headers={"API-Key": api_key}
    )
    return response.json() 
# response = requests.get(
#     "https://amazon-scraper-api.omkar.cloud/amazon/products/category",
#     params={"category_id": "16225007011", "country_code": "IN"},
#     headers={"API-Key": OC_API_KEY}
# )

# print(response.json())
```
